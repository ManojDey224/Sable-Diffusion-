```python
# Import necessary libraries
import torch
from torch import autocast
from diffusers import StableDiffusionPipeline, LMSDiscreteScheduler
from PIL import Image

# Function to generate images
def generate_images(prompt, num_images=1, guidance_scale=7.5, steps=50):
    """
    Generates images based on the given prompt using Stable Diffusion model.

    Parameters:
    - prompt (str): The text prompt for image generation.
    - num_images (int): Number of images to generate.
    - guidance_scale (float): Guidance scale for image generation.
    - steps (int): Number of diffusion steps.

    Returns:
    - List of generated images.
    """

    # Set up the scheduler and pipeline
    scheduler = LMSDiscreteScheduler(beta_start=0.00085, beta_end=0.012, beta_schedule="scaled_linear")
    model_id = "CompVis/stable-diffusion-v1-4"
    device = "cuda"  # Use GPU

    # Load the pipeline
    pipe = StableDiffusionPipeline.from_pretrained(model_id, scheduler=scheduler, torch_dtype=torch.float16)
    pipe = pipe.to(device)

    # Generate images
    generated_images = []
    with autocast("cuda"):
        for _ in range(num_images):
            image = pipe(prompt, guidance_scale=guidance_scale, num_inference_steps=steps)["images"][0]
            generated_images.append(image)

    return generated_images

# Function to save images
def save_images(images, output_dir="output_images"):
    """
    Saves the generated images to the specified directory.

    Parameters:
    - images (list): List of images to save.
    - output_dir (str): Directory to save the images.
    """
    import os

    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    for i, image in enumerate(images):
        image_path = os.path.join(output_dir, f"generated_image_{i+1}.png")
        image.save(image_path)
        print(f"Image saved at {image_path}")

# Example usage
if __name__ == "__main__":
    prompt = "A highly detailed, realistic portrait of a golden retriever-like dog with golden brown fur, expressive eyes, and a gentle, soulful expression. The background is dark and minimalist, emphasizing the dog's features. Painted in a hyper-realistic digital art style."
    num_images = 3
    images = generate_images(prompt, num_images)

    # Save the generated images
    save_images(images)
```
